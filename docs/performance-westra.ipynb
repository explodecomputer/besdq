# BESDQ Performance — Westra eQTL

Benchmarks against the Westra et al. (2013) blood eQTL dataset (506,049 SNPs × 5,966 probes).
This notebook is executed automatically by CI on every push to `main`.

```python
%%bash
smr="../data/smr"
myeqtl="../data/westra_eqtl_hg19"
# Use SMR
time $smr \
    --beqtl-summary $myeqtl \
    --query 5e-5 \
    --snp-chr 1 \
    --from-snp-kb 100 \
    --to-snp-kb 2000 \
    --probe-chr 1 \
    --from-probe-kb 1000 \
    --to-probe-kb 2000 \
    --out results/myquery_smr
```

```python
%%bash
myeqtl="../data/westra_eqtl_hg19"
# Use besdq
pip install -e ..
time python3 -m besdq.cli \
    --besd $myeqtl \
    --query 5.0e-4 \
    --snp-chrpos 1:1191870 \
    --probe-chrpos 1:1140818 \
    --out results/remyquery
```

```python
%%bash
myeqtl="../data/westra_eqtl_hg19"

# Use installed package
time besdq \
    --besd $myeqtl \
    --query 5e-4 \
    --snp-chr 1 \
    --from-snp-kb 100 \
    --to-snp-kb 2000 \
    --probe-chr 1 \
    --from-probe-kb 1000 \
    --to-probe-kb 2000 \
    --out results/myquery
```
